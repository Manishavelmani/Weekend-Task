#### Split the dataset x and y

```python
X = data.drop(columns=["Churn"])
y = data["Churn"]
```

```python
print("X shape:",X.shape)
print("y shape:",y.shape)
```

#### the target column is imbalanced.

#### Train/Test Split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
import joblib
joblib.dump(X_train, "../models/X_train_raw.pkl")
joblib.dump(X_test, "../models/X_test_raw.pkl")
```

#### Numerical Features

```python
numerical_features = [
    "SeniorCitizen",
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]
```

#### Categorical Features

```python
categorical_features = [
    "gender",
    "Partner",
    "Dependents",
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaperlessBilling",
    "PaymentMethod"
]
```

```python
print("Numerical features:")
print(numerical_features)
print("\nCategorical features:")
print(categorical_features)
```

```python
print("Total features:", len(numerical_features) + len(categorical_features))
print("X_train columns:", X_train.shape[1])
```

#### One-hot encoding(categorical) and scaling(numerical)

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

#### Create the preprocessing transformer

```python
preprocessor = ColumnTransformer(
    transformers=[("num",StandardScaler(),numerical_features),
                  ("cat",OneHotEncoder(handle_unknown="ignore",sparse_output=False),
                   categorical_features
                  )])
```

#### Fit only on training data

```python
X_train_processed=preprocessor.fit_transform(X_train)
X_test_processed=preprocessor.transform(X_test)
```

```python
print("X_train original shape:", X_train.shape)
print("X_train processed shape:", X_train_processed.shape)
print("X_test original shape:", X_test.shape)
print("X_test processed shape:", X_test_processed.shape)
```

```python
print(type(X_train_processed))
print(type(X_test_processed))
```

#### Check and handle outliers

```python
numerical_features = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]
```

```python
for col in ["tenure", "MonthlyCharges", "TotalCharges"]:
    Q1=X_train[col].quantile(0.25)
    Q3=X_train[col].quantile(0.75)
    IQR=Q3-Q1
    lower_bound=Q1-1.5*IQR
    upper_bound=Q3+1.5*IQR
    outliers=X_train[(X_train[col]<lower_bound)|(X_train[col]>upper_bound)]
    print(f"{col}")
    print("Lower bound:", lower_bound)
    print("Upper bound:", upper_bound)
    print("Outlier count:", len(outliers))
    print()
```

#### No outliers in the datasets

```python
print("Negative tenure:", (X_train["tenure"] < 0).sum())
print("Negative MonthlyCharges:", (X_train["MonthlyCharges"] < 0).sum())
print("Negative TotalCharges:", (X_train["TotalCharges"] < 0).sum())
```
