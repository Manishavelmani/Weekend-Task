#### Select The Text Columns

```python
data["Text"] = (data["Ticket Subject"].fillna("")+ " "+ 
                data["Ticket Description"].fillna(""))
```

#### Text Statistics

```python
#Number of characters
data["Text_Chars"] = data["Text"].str.len()
```

```python
#Number of words
data["Text_Word_Count"] = (data["Text"].str.split().str.len())
```

```python
print(data["Text_Chars"].describe())
print(data["Text_Word_Count"].describe())
```

#### Clean the text

```python
import re
def clean_text(text):
    text = str(text).lower()
    # Remove URLs
    text = re.sub(r"http\S+|www\S+", "", text)
    # Keep letters, spaces, and apostrophes
    text = re.sub(r"[^a-z\s']", " ", text)
    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()
    return text
```

```python
data["Clean_Text"] = data["Text"].apply(clean_text)
```

```python
print(data["Clean_Text"].iloc[0])
```

```python
data[["Text", "Clean_Text"]].head()
```

Tokenization

```python
data["Tokens"] = data["Clean_Text"].str.split()
```

```python
print(data["Tokens"].iloc[0])
```

#### Stop-word handling

```python
import nltk
from nltk.corpus import stopwords
nltk.download("stopwords")
stop_words = set(stopwords.words("english"))
```

```python
data["Tokens_No_Stopwords"] = data["Tokens"].apply(
    lambda tokens: [
        word for word in tokens
        if word not in stop_words
    ]
)
```

```python
print(data["Tokens_No_Stopwords"].iloc[0])
```

#### Word frequency

```python
from collections import Counter
all_words = [
    word
    for tokens in data["Tokens_No_Stopwords"]
    for word in tokens
]
word_counts = Counter(all_words)
word_frequency = pd.DataFrame(word_counts.most_common(20),columns=["Word", "Frequency"])
word_frequency
```

#### Bigram analysis

```python
from sklearn.feature_extraction.text import CountVectorizer
bigram_vectorizer = CountVectorizer(
    ngram_range=(2, 2),
    stop_words="english",
    token_pattern=r"(?u)\b[a-zA-Z']+\b"
)
bigram_matrix = bigram_vectorizer.fit_transform(data["Clean_Text"])
bigram_counts = pd.DataFrame({
    "Bigram": bigram_vectorizer.get_feature_names_out(),
    "Frequency": bigram_matrix.sum(axis=0).A1
}).sort_values("Frequency",ascending=False)
bigram_counts.head(20)
```

#### Trigram Analysis

```python
trigram_vectorizer = CountVectorizer(
    ngram_range=(3, 3),
    stop_words="english",
    token_pattern=r"(?u)\b[a-zA-Z']+\b"
)
trigram_matrix = trigram_vectorizer.fit_transform(data["Clean_Text"])
trigram_counts = pd.DataFrame({
    "Trigram": trigram_vectorizer.get_feature_names_out(),
    "Frequency": trigram_matrix.sum(axis=0).A1
}).sort_values("Frequency",ascending=False)
trigram_counts.head(20)
```

```python
X = data["Clean_Text"]
y = data["Ticket Type"]
```

```python
print(y.value_counts())
print("\nNumber of classes:", y.nunique())
```

#### Train/Test Split

```python
from sklearn.model_selection import train_test_split
X_train_text, X_test_text, y_train_text, y_test_text = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training samples:", len(X_train_text))
print("Testing samples :", len(X_test_text))
```

```python
print("\nTraining target distribution:")
print(y_train_text.value_counts())
```

#### Create TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(stop_words="english",ngram_range=(1, 2),max_features=20000)
```

#### Fit TF-IDF only on training text

```python
X_train_tfidf = tfidf.fit_transform(X_train_text)
X_test_tfidf = tfidf.transform(X_test_text)
```

#### Check the TF-IDF matrix

```python
print("X_train TF-IDF shape:", X_train_tfidf.shape)
print("X_test TF-IDF shape :", X_test_tfidf.shape)
```

#### Inspect TF-IDF vocabulary

```python
print("Number of vocabulary terms:",len(tfidf.vocabulary_))
```

```python
feature_names = tfidf.get_feature_names_out()
print(feature_names[:20])
```

## Traditional NLP Model

#### 1.Logistic Expression

```python
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
import time
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
```

```python
text_lr = LogisticRegression(max_iter=1000,random_state=42)
```

```python
start_time = time.time()
text_lr.fit(X_train_tfidf,y_train_text)
training_time = time.time() - start_time
print("Training time:", training_time)
```

```python
y_pred_text_lr = text_lr.predict(X_test_tfidf)
y_proba_text_lr = text_lr.predict_proba(X_test_tfidf)
```

#### Evaluate the metrics

```python
text_lr_accuracy = accuracy_score(y_test_text,y_pred_text_lr)
text_lr_precision = precision_score(y_test_text,y_pred_text_lr,average="weighted",zero_division=0)
text_lr_recall = recall_score(y_test_text,y_pred_text_lr,average="weighted",zero_division=0)
text_lr_f1 = f1_score(y_test_text,y_pred_text_lr,average="weighted",zero_division=0)
```

```python
nlp_results = []
nlp_results.append({
    "Model": "TF-IDF + Logistic Regression",
    "Accuracy": text_lr_accuracy,
    "Precision": text_lr_precision,
    "Recall": text_lr_recall,
    "F1-Score": text_lr_f1,
    "Training Time (sec)": training_time
})
```

```python
nlp_results
```

```python
nlp_results_df = pd.DataFrame(nlp_results).round(4)
nlp_results_df
```

```python
nlp_results.append({
    "Model": "TF-IDF + Naive Bayes",
    "Accuracy": text_nb_accuracy,
    "Precision": text_nb_precision,
    "Recall": text_nb_recall,
    "F1-Score": text_nb_f1,
    "Training Time (sec)": training_time
})
```
