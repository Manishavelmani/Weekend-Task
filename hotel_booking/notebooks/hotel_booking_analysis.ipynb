## Outlier Detection – IQR Method

The Interquartile Range (IQR) method was used to identify potential outliers in the `adr` (Average Daily Rate) feature. Values outside the lower and upper bounds (Q1 − 1.5 × IQR and Q3 + 1.5 × IQR) were flagged as potential outliers for further investigation.

```python
# Calculate Quartiles
Q1 = df["adr"].quantile(0.25)
Q3 = df["adr"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate Bounds
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
adr_outliers = df[
    (df["adr"] < lower_bound) |
    (df["adr"] > upper_bound)
]

print("Number of ADR Outliers:", len(adr_outliers))
```

```python
plt.figure(figsize=(8,6))

sns.boxplot(y=df["adr"], color="skyblue")

plt.title("Box Plot of Average Daily Rate (ADR)", fontsize=14)
plt.ylabel("ADR")

plt.show()                                         
```

## Box-Cox Transformation

```python
from scipy.stats import boxcox

df_boxcox = df.copy()

df_boxcox["lead_time_boxcox"], lambda_value = boxcox(
    df_boxcox["lead_time"] + 1
)

print("Lambda Value:", lambda_value)

df_boxcox[["lead_time", "lead_time_boxcox"]].head()
```

# Probability Analysis

Probability measures the likelihood of an event occurring.

Formula:

Probability = Number of Desired Outcomes / Total Number of Outcomes

In this analysis, we calculate the probability that a booking is cancelled.

```python
total_bookings = len(df)

cancelled_bookings = df["is_canceled"].sum()

probability_cancelled = cancelled_bookings / total_bookings

print("Total Bookings :", total_bookings)
print("Cancelled Bookings :", cancelled_bookings)
print("Probability of Cancellation :", round(probability_cancelled,3))
```

```python
plt.figure(figsize=(6,5))

sns.countplot(
    data=df,
    x="is_canceled",
    hue="is_canceled",
    palette="Set2",
    legend=False
)

plt.title("Booking Cancellation")

plt.xticks([0,1],["Not Cancelled","Cancelled"])

plt.show()
```

## Conditional Probability

Conditional Probability measures the probability of an event occurring given that another event has already occurred.

In this analysis, we calculate the probability of a booking being cancelled given that the booking belongs to a City Hotel.

```python
city_hotel = df[df["hotel"] == "City Hotel"]

city_cancelled = city_hotel["is_canceled"].sum()

probability = city_cancelled / len(city_hotel)

print("City Hotel Bookings:", len(city_hotel))
print("Cancelled:", city_cancelled)
print("Conditional Probability:", round(probability,3))
```

## Central Limit Theorem (CLT)

The Central Limit Theorem states that the sampling distribution of the sample mean approaches a normal distribution as the sample size increases, regardless of the original population distribution.

In this project, multiple samples of ADR values are selected from hotel booking data. The mean ADR of each sample is calculated to demonstrate CLT.

```python
sampling_mean = sampling_df["sample_mean_adr"].mean()

sampling_std = sampling_df["sample_mean_adr"].std()

print("Mean of Sample Means:", round(sampling_mean,2))
print("Standard Deviation of Sample Means:", round(sampling_std,2))
```

```python
plt.figure(figsize=(10,5))

sns.histplot(
    data=sampling_df,
    x="sample_mean_adr",
    bins=20,
    kde=True
)

plt.axvline(
    sampling_mean,
    linestyle="--",
    label="Mean of Sample Means"
)

plt.title("Central Limit Theorem - Sampling Distribution of ADR")

plt.xlabel("Sample Mean ADR")

plt.ylabel("Frequency")

plt.legend()
plt.savefig("../static/images/clt_distribution.png", dpi=300, bbox_inches="tight")

plt.show()
```

## Central Limit Theorem Result Interpretation

The sampling distribution of ADR was created by taking multiple random samples from the hotel booking dataset.

The mean of the sample means is 106.42, which represents the estimated population average ADR.

The standard deviation of sample means is 2.89, indicating that sample averages have low variation and are centered around the population mean.

This demonstrates the Central Limit Theorem, where sample means form an approximately normal distribution even when the original data contains variations and outliers.

## Confidence Interval Analysis

A confidence interval provides a range of values that is likely to contain the true population parameter.

In this project, a 95% confidence interval is calculated for the average ADR using sample booking data.

```python
sample = df["adr"].sample(
    n=1000,
    random_state=42
)

sample_mean = sample.mean()

print("Sample Mean ADR:", round(sample_mean,2))
```

```python
from scipy import stats

confidence_level = 0.95

confidence_interval = stats.t.interval(
    confidence_level,
    len(sample)-1,
    loc=sample_mean,
    scale=stats.sem(sample)
)

print("95% Confidence Interval:")
print(confidence_interval)
```

```python
plt.figure(figsize=(8,5))

plt.errorbar(
    x=["ADR"],
    y=[sample_mean],
    yerr=[
        [
            sample_mean - confidence_interval[0]
        ],
        [
            confidence_interval[1] - sample_mean
        ]
    ],
    fmt="o"
)

plt.title("95% Confidence Interval for ADR")

plt.ylabel("Average Daily Rate")
plt.savefig("../static/images/confidence_interval.png", dpi=300, bbox_inches="tight")

plt.show()
```

## Normality Test - Shapiro-Wilk Test

The Shapiro-Wilk test is used to determine whether a dataset follows a normal distribution.

Null Hypothesis (H0):
The data is normally distributed.

Alternative Hypothesis (H1):
The data is not normally distributed.

A significance level of 0.05 is used for decision making.

```python
from scipy.stats import shapiro

adr_sample = df["adr"].sample(
    n=5000,
    random_state=42
)

stat, p_value = shapiro(adr_sample)

print("Test Statistic:", stat)
print("P-value:", p_value)
```

```python
if p_value > 0.05:
    print("ADR data is normally distributed")
else:
    print("ADR data is not normally distributed")
```

```python
import scipy.stats as stats
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

stats.probplot(
    adr_sample,
    dist="norm",
    plot=plt
)

plt.title("QQ Plot - ADR")
plt.savefig("../static/images/normal_distribution.png", dpi=300, bbox_inches="tight")

plt.show()
```

## Normality Test Result

The Shapiro-Wilk test was performed on the ADR feature to check whether it follows a normal distribution.

The p-value obtained was significantly less than 0.05.

# Independent t-Test

The Independent t-Test is used to compare the average ADR between City Hotel and Resort Hotel.

## Null Hypothesis (H₀)

The average ADR of City Hotel and Resort Hotel is equal.

## Alternative Hypothesis (H₁)

The average ADR of City Hotel and Resort Hotel is different.

Since the Levene Test indicated unequal variances, Welch's t-Test (equal_var=False) is used.

Significance Level (α) = 0.05

```python
from scipy.stats import ttest_ind

city_adr = df[df["hotel"] == "City Hotel"]["adr"]

resort_adr = df[df["hotel"] == "Resort Hotel"]["adr"]

t_stat, p_value = ttest_ind(
    city_adr,
    resort_adr,
    equal_var=False
)

print("T-Statistic:", t_stat)
print("P-value:", p_value)
```

```python
if p_value < 0.05:
    print("Reject Null Hypothesis")
    print("There is a significant difference in ADR between City Hotel and Resort Hotel.")
else:
    print("Fail to Reject Null Hypothesis")
    print("No significant difference in ADR between City Hotel and Resort Hotel.")
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="hotel",
    y="adr",
    hue="hotel",
    palette="Set2",
    legend=False
)

plt.title("ADR Comparison Between Hotel Types")
plt.xlabel("Hotel Type")
plt.ylabel("Average Daily Rate (ADR)")

plt.show()
```

The Independent t-Test showed a statistically significant difference in the Average Daily Rate (ADR) between City Hotels and Resort Hotels.

This indicates that hotel type significantly influences room pricing. Resort Hotels and City Hotels follow different pricing strategies based on factors such as customer demand, seasonality, and location.

This insight can help hotel management optimize pricing policies for different hotel categories.

# One-Way ANOVA

One-Way ANOVA is used to compare the average ADR among different customer types.

## Null Hypothesis (H₀)

The average ADR is the same for all customer types.

## Alternative Hypothesis (H₁)

At least one customer type has a different average ADR.

Significance Level (α) = 0.05

```python
from scipy.stats import f_oneway
```

```python
transient = df[df["customer_type"] == "Transient"]["adr"]

contract = df[df["customer_type"] == "Contract"]["adr"]

group = df[df["customer_type"] == "Group"]["adr"]

party = df[df["customer_type"] == "Transient-Party"]["adr"]
```

```python
f_stat, p_value = f_oneway(
    transient,
    contract,
    group,
    party
)

print("F-Statistic:", f_stat)
print("P-value:", p_value)
```

```python
if p_value < 0.05:
    print("Reject Null Hypothesis")
    print("Average ADR differs among customer types.")
else:
    print("Fail to Reject Null Hypothesis")
    print("Average ADR is similar among customer types.")
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x="customer_type",
    y="adr",
    hue="customer_type",
    palette="Set2",
    legend=False
)

plt.title("ADR Distribution by Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Average Daily Rate (ADR)")

plt.xticks(rotation=20)

plt.show()
```

## One-Way ANOVA Result

A One-Way ANOVA test was performed to compare the Average Daily Rate (ADR) among different customer types.

The p-value obtained was less than 0.05.

Therefore, the null hypothesis was rejected.

This indicates that at least one customer type has a significantly different average ADR.

The result suggests that customer type influences hotel pricing. Hotels may apply different pricing strategies for Transient, Contract, Group, and Transient-Party customers.

# Tukey HSD Post-hoc Test

The One-Way ANOVA test indicated a significant difference in Average Daily Rate (ADR) among customer types.

To determine which specific customer types differ significantly, Tukey's Honestly Significant Difference (HSD) post-hoc test was performed.

The test compares all possible pairs of customer types while controlling the overall Type I error rate.

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd
```

```python
tukey = pairwise_tukeyhsd(
    endog=df["adr"],
    groups=df["customer_type"],
    alpha=0.05
)

print(tukey)
```

```python
tukey.plot_simultaneous(figsize=(10,6))

plt.title("Tukey HSD Comparison of ADR by Customer Type")

plt.show()
```

## Tukey HSD Post-hoc Test Result

Since the One-Way ANOVA test indicated a significant difference in ADR among customer types, Tukey's HSD post-hoc test was performed.

The results showed significant ADR differences between most customer type pairs.

However, there was no statistically significant difference between **Group** and **Transient-Party** customers (p = 0.5633).

This indicates that these two customer segments have similar pricing, while the remaining customer types follow different pricing strategies.

# Chi-Square Test

The Chi-Square Test is used to determine whether hotel type and booking cancellation are associated.

## Null Hypothesis (H₀)

Hotel type and cancellation status are independent.

## Alternative Hypothesis (H₁)

Hotel type and cancellation status are associated.

Significance Level (α) = 0.05

```python
from scipy.stats import chi2_contingency
```

```python
contingency_table = pd.crosstab(
    df["hotel"],
    df["is_canceled"]
)

print(contingency_table)
```

```python
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-Square Statistic:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)
```

```python
if p_value < 0.05:
    print("Reject Null Hypothesis")
    print("Hotel type and cancellation are associated.")
else:
    print("Fail to Reject Null Hypothesis")
    print("Hotel type and cancellation are independent.")
```

## Chi-Square Test Result

A Chi-Square Test of Independence was performed to examine the relationship between hotel type and booking cancellation.

The test produced a Chi-Square statistic of 454.68 with a p-value less than 0.05.

Therefore, the null hypothesis was rejected.

This indicates that hotel type and booking cancellation are significantly associated. Cancellation behavior differs between City Hotels and Resort Hotels.

Hotel managers can use this insight to implement different cancellation policies, pricing strategies, and booking management practices for different hotel types.
