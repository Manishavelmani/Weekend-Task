Handle Outlires in numerical coulmns in the datsets

First check Outliers in Salesprice 

```python
Q1=data["SalePrice"].quantile(0.25)
Q3=data["SalePrice"].quantile(0.75)
IQR=Q3-Q1
lower_bound=Q1-(1.5*IQR)
upper_bound=Q3+(1.5*IQR)
SalePrice_Outliiers=data[
    (data["SalePrice"]<lower_bound)|(data["SalePrice"]>upper_bound)
]
print("Number of Oulers:" ,len(SalePrice_Outliiers))
```

```python
print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)
print("Lower Bound:",lower_bound)
print("Upper Bound:",upper_bound)
print("Number of Outliers:", len(SalePrice_Outliiers))
```

Using the IQR method, 61 outliers were detected in the SalePrice column. These correspond to houses with sale prices above the upper bound of 340,037.5. Since these values likely represent genuine high-value properties rather than data entry errors, they should be investigated further before deciding whether to remove or transform them.

Find Oulires fo all numerical coulmns

```python
outlier_summary = {}
for col in numerical_col:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 -(1.5*IQR)
    upper_bound = Q3 +(1.5*IQR)
    outliers = data[(data[col] < lower_bound) | (data[col] > upper_bound)]
    outlier_summary[col] =len(outliers)
    outlier_count = ((data[col] < lower_bound) | (data[col] > upper_bound)).sum()

    outlier_df = pd.DataFrame(outlier_summary.items(), columns=["Column","Outlier Count"])

print(outlier_df.sort_values(by="Outlier Count", ascending=False))
```

Phase 4 – Model Development

1.Import Libraries 

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import(mean_absolute_error,mean_squared_error,r2_score)
```

2.prepare input(features) and output(target)

```python
data=data.drop("Id", axis=1)
```

Remove the identifier column in the datasets

```python
data.head()
```

```python
x=data.drop("SalePrice",axis=1)
y=data["SalePrice"]
```

x - input features
#y - Saleprice(target)

Encode the categorical variables because Machine Learning Algoring dont Undersztand Text

```python
x=pd.get_dummies(x,drop_first=True,dtype=int)
```

```python
x.info()
```

```python
bool_cols = x.select_dtypes(include="bool").columns
x[bool_cols] = x[bool_cols].astype(int)
```

```python
x.info()
```

```python
x.head()
```

```python
x["GarageYrBlt"] = x["GarageYrBlt"].fillna(0)
```

```python
print(x.isnull().sum().sum())
```

Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42
)
```

Linear Regression

```python
from sklearn.linear_model import LinearRegression
```

create a model for train the features

```python
model=LinearRegression()
#train the model
model.fit(X_train,y_train)
#make prediction
y_pred=model.predict(X_test)
#view prediction
print(y_pred[:10])
```

predict the first 10 house price using linear regression

Next Step:Evaluate the model

calculate the metrics

```python
mae=mean_absolute_error(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
rmse=np.sqrt(mse)
r2=r2_score(y_test,y_pred)
```

print results

```python
print("Linear Regression Results")
print("MAE :", mae)
print("MSE :", mse)
print("RMSE :", rmse)
print("R2 Score :", r2)
```

Linear Regression achieved an R² score of 0.8509, indicating that the model explains approximately 85% of the variation in house prices. The average prediction error (MAE) was around 20,438, showing that the model provides reasonably accurate predictions. These results make Linear Regression a strong baseline model for comparison with more advanced regression algorithms.
