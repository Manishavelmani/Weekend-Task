```python
import pandas as pd

reviews = [

    "I absolutely loved this movie. The acting was brilliant and the story was engaging.",

    "This was one of the worst movies I have ever watched. The plot was boring and confusing.",

    "The movie was entertaining from beginning to end. I would definitely watch it again.",

    "I was disappointed by the weak storyline and poor acting. It was difficult to stay interested.",

    "The characters were well developed and the performances were excellent.",

    "Nothing about this movie worked for me. The dialogue was bad and the ending was awful.",

    "A wonderful movie with beautiful visuals, great music, and fantastic performances.",

    "The movie started well but became boring after a while. The ending was also disappointing.",

    "I enjoyed every scene of this film. The story was emotional and the actors did a great job.",

    "This movie was a complete waste of time. The story was predictable and the acting was terrible."

]

df = pd.DataFrame({
    "Review": reviews
})

df.to_csv("reviews_2.csv", index=False)

print("reviews_2.csv created successfully!")
```
